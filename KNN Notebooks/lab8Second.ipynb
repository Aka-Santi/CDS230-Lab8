```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('instagram_global_influencers_2026 (1).csv')
```

```python
# Print the feature names
print(df.columns)

# Print the top five records
print(df.head())
```

```python
# Select features
X = df[['influence_score', 'posts', 'followers',
        'avg_likes', '60_day_eng_rate',
        'new_post_avg_likes', 'total_likes']]

# Select the target
y = df['category']
```

```python
# Print data (feature) shape
print(X.shape)

# Print target (label) shape
print(y.shape)
```

```python
# Import train_test_split function
from sklearn.model_selection import train_test_split

# Split dataset into training set and test set
# 70% training and 30% test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
# Import StandardScaler
from sklearn.preprocessing import StandardScaler

# Scale the features
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# Import K-nearest neighbors classifier model
from sklearn.neighbors import KNeighborsClassifier

# Create KNN Classifier with K=5
knn = KNeighborsClassifier(n_neighbors=5)
```

```python
# Train the model using the training sets
knn.fit(X_train, y_train)
```

```python
# Predict the response for test dataset
y_pred = knn.predict(X_test)
```

```python
# Import metrics
from sklearn import metrics

# Model accuracy
print("Accuracy:", metrics.accuracy_score(y_test, y_pred))
```

```python
# Import metrics
from sklearn.metrics import precision_score, recall_score, f1_score

# Precision
precision = precision_score(y_test, y_pred, average='weighted',
                            zero_division=0)
print("Precision:", precision)

# Recall
recall = recall_score(y_test, y_pred, average='weighted',
                      zero_division=0)
print("Recall:", recall)

# F1-score
f1 = f1_score(y_test, y_pred, average='weighted',
              zero_division=0)
print("F1-score:", f1)
```

I used KNN to predict several categories of people's instagram accounts. the model used their followers, posts, and likes to make an accurate predictions.
70% of the data is used this time to train and 30%of the data is used for testing. the model compares similar insta accounts to make a prediction their categories.
the results show how the model can make an accurate prediction regarding instagram account categories.
